# Langkah 1 — Menangani nilai hilang (imputasi median/modus).

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler
data = {
    'usia': [25, 30, np.nan, 45, 35, np.nan, 50, 23],
    'pendapatan': [5000000, 7500000, 10000000, np.nan, 8000000, 6000000, 12000000, np.nan],
    'pendidikan': ['SMA', 'S1', 'S2', 'S1', 'SMA', 'S2', 'S1', 'SMA'],
    'kota': ['Jakarta', 'Bandung', 'Jakarta', 'Surabaya', 'Bandung', 'Jakarta', 'Surabaya', 'Bandung'],
    'membeli': ['Tidak', 'Ya', 'Ya', 'Tidak', 'Ya', 'Tidak', 'Ya', 'Tidak']
}

df = pd.DataFrame(data)

df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
print(df)
```

4. Jelaskan mengapa fit_transform pada latih tetapi hanya transform pada uji.
Alasannya : fit_transform() pada data latih bertindak untuk menghitung parameter statistik (seperti rata-rata dan standar deviasi untuk StandardScaler, atau nilai min dan max untuk MinMaxScaler) sekaligus mengaplikasikannya pada data latih. Pada data uji, kita hanya boleh menggunakan transform() dengan memanfaatkan parameter statistik yang sudah dipelajari dari data latih. Hal ini wajib dilakukan guna mencegah data leakage (kebocoran informasi dari data uji ke data latih), sehingga model mensimulasikan kondisi pengujian aktual terhadap data yang belum pernah dilihat sebelumnya.

# E. Refleksi

1. Mengapa urutan 'split dulu, baru scaling' penting?

Jawab :Membagi data sebelum scaling (train-test split terlebih dahulu) sangat krusial untuk mencegah data leakage. Jika kita melakukan scaling pada seluruh dataset sebelum pembagian, informasi statistik dari data uji (seperti nilai rata-rata, minimum, atau maksimum) akan ikut terbawa ke dalam data latih. Hal ini membuat evaluasi model menjadi terlalu optimistis (bias) karena model secara tidak langsung sudah "mengenal" karakteristik data uji.

2. Kapan label encoding dan kapan one-hot encoding?

Label Encoding / Ordinal Encoding: Gunakan ketika variabel kategorikal memiliki tingkatan atau urutan logis (data ordinal), misalnya jenjang pendidikan (SMA < S1 < S2) atau tingkat kepuasan (Rendah, Sedang, Tinggi).   
Sedangkan One-Hot Encoding: Gunakan ketika variabel kategorikal tidak memiliki urutan atau tingkatan (data nominal), misalnya nama kota (Jakarta, Bandung, Surabaya) atau status jenis kelamin. Menggunakan label encoding pada data nominal dapat menyesatkan algoritma karena menganggap ada urutan matematis pada kategori tersebut.

3. Apa perbedaan normalisasi dan standardisasi?

Normalisasi mengubah nilai data ke rentang tertentu , biasanya 0 sampai 1 . Contohnya menggunakan MinMaxScaler.
Sedangkan Standardisasi mengubah data berdasarkan rata-rata (mean) dan standar deviasi ,0 dan standar deviasi 1 . Contohnya menggunakan StandardScaler.
